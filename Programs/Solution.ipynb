```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\namra\OneDrive\Documents\Python-Journey\Programs\company_sales_data.csv")
print(df.head())
```

```python
#q1
month = df["month_number"]
profit = df["total_profit"]
plt.plot(month,profit)
plt.xlabel("Month Number")
plt.ylabel("Total profit")
plt.show()
```

```python
#q2
month = df["month_number"]
profit = df["total_profit"]
plt.plot(month, profit, linestyle=':', color='red',marker='o', markerfacecolor='black', markeredgecolor='black', linewidth=2, label="Profit data of last year")
plt.xlabel("Month Number")
plt.ylabel("Sold units number")
plt.title("Company sales of last year")
plt.legend(loc="lower right")
plt.show()
```

```python
#q3
month = df["month_number"]
plt.plot(month, df["facecream"], label="Face Cream")
plt.plot(month, df["facewash"], label="Face Wash")
plt.plot(month, df["toothpaste"], label="Toothpaste")
plt.plot(month, df["bathingsoap"], label="Bathing Soap")
plt.plot(month, df["shampoo"], label="Shampoo")
plt.plot(month, df["moisturizer"], label="Moisturizer")

plt.xlabel("Month Number")
plt.ylabel("Units Sold")

plt.title("Product Sales Data")

plt.legend()

plt.show()
```

```python
#q4
month = df["month_number"]
toothpaste = df["toothpaste"]
plt.scatter(month, toothpaste)
plt.xlabel("Month Number")
plt.ylabel("Toothpaste Sales")
plt.title("Toothpaste Sales per Month")
plt.show()
```

```python
#q5
month = df["month_number"]
plt.bar(month-0.2, df["facecream"], width=0.4, label="Face Cream")
plt.bar(month+0.2, df["facewash"],  width =0.4, label = "Face Wash")
plt.xlabel("Month Number")
plt.ylabel("Units Sold")
plt.title("Face Cream and Face Wash Sales")
plt.legend()
plt.show()
```

```python
#q6
month = df["month_number"]
bathing_soap = df["bathingsoap"]
plt.bar(month, bathing_soap)
plt.xlabel("Month Number")
plt.ylabel("Bathing soap sales")
plt.title("Bathing Soap Sales per Month")
plt.show()
```

```python
#q7-a
profit = df["total_profit"]
plt.hist(profit, bins =5)
plt.xlabel("Total Profit")
plt.ylabel("Frequency")
plt.title("Distribution of Total Profit")
plt.show()
```

```python
#q7-b
plt.bar(month, profit)
plt.xlabel("Month")
plt.ylabel("Profit")
plt.title("Monthly Profit")
plt.show()
```

```python
#q8
products = [
    "facecream",
    "facewash",
    "toothpaste",
    "bathingsoap",
    "shampoo",
    "moisturizer"
]

total_sales = [
    df["facecream"].sum(),
    df["facewash"].sum(),
    df["toothpaste"].sum(),
    df["bathingsoap"].sum(),
    df["shampoo"].sum(),
    df["moisturizer"].sum()
]

plt.pie(
    total_sales,
    labels=products,
    autopct="%1.1f%%"
)

plt.title("Total Sales of Each Product")

plt.show()
```

```python
#q9
# First subplot
plt.subplot(1, 2, 1)

plt.plot(month, df["bathingsoap"])

plt.xlabel("Month Number")
plt.ylabel("Sales")

plt.title("Bathing Soap Sales")


# Second subplot
plt.subplot(1, 2, 2)

plt.plot(month, df["facewash"])

plt.xlabel("Month Number")
plt.ylabel("Sales")

plt.title("Face Wash Sales")


plt.tight_layout()
plt.show()
```
